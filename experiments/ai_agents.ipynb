```python
import ollama

response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "user",
            "content": (
                "A notebook costs ₹100. "
                "A pen costs ₹20 less than the notebook. "
                "Together, they cost ₹120. "
                "Is this possible? Explain your reasoning."
            )
        }
    ],
    think=True,
    stream=False
)

print("REASONING:")
print(response.message.thinking)

print("\nFINAL ANSWER:")
print(response.message.content)
```

```python
from agents import (
    Agent,
    Runner,
    OpenAIChatCompletionsModel,
    set_tracing_disabled,
)
from openai import AsyncOpenAI

# Disable OpenAI-hosted tracing for this local experiment.
set_tracing_disabled(True)

# Connect to Ollama's local OpenAI-compatible endpoint.
ollama_client = AsyncOpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama",  # Placeholder; not an OpenAI API key.
)

# Configure the local model.
ollama_model = OpenAIChatCompletionsModel(
    model="qwen3:0.6b",
    openai_client=ollama_client,
)

# Create an agent.
agent = Agent(
    name="Local Qwen Agent",
    instructions="You are a helpful assistant. Answer clearly and concisely.",
    model=ollama_model,
)
```

```python
import asyncio

result = await Runner.run(
    agent,
    "You are such a loser. "
)

print(result.final_output)
```

```python
from agents import function_tool


@function_tool
def calculate_score(number: int) -> int:
    """
    Calculate score by multiplying the input number by 100.
    """
    return number * 100
```

```python
agent = Agent(
    name="Score Agent",
    instructions="Use the calculate_score tool when the user asks you to calculate a score.",
    model=ollama_model,
    tools=[calculate_score],
)
```

```python
result = await Runner.run(
    agent,
    "Calculate the score for 7."
)

print(result.final_output)
```

```python
import time

start = time.perf_counter()

result = await Runner.run(
    agent,
    "Calculate the score for 7."
)

end = time.perf_counter()

print(f"Total time: {end - start:.2f}s")
print(result.final_output)
```

```python
import ollama
import time

start = time.perf_counter()

response = ollama.chat(
    model="qwen3:0.6b",
    messages=[
        {
            "role": "user",
            "content": "Say hello."
        }
    ],
    think=False,
    stream=False
)

elapsed = time.perf_counter() - start

print(f"Time: {elapsed:.2f}s")
print(response.message.content)
```

```python
for i in range(2):

    start = time.perf_counter()

    response = ollama.chat(
        model="qwen3:0.6b",
        messages=[
            {"role": "user", "content": "Say hello."}
        ],
        think=False,
        stream=False,
        keep_alive=-1
    )

    elapsed = time.perf_counter() - start

    print(f"Run {i + 1}: {elapsed:.2f}s")
```
